### Análisis de Rendimiento Goleador según la Posición

El análisis de los datos históricos revela cómo la evolución táctica de Messi a lo largo de su carrera ha impactado directamente en sus registros goleadores:

1.  Falso 9: Aunque tradicionalmente se le asocia con la banda derecha en sus inicios, el periodo donde jugó como falso 9 coincide con sus temporadas más explosivas y prolíficas de cara al arco, incluyendo la histórica temporada 2011/2012. En esta pocisión, Messi maximiza su frecuencia goleadora al estar más cerca del área y ser el finalizador absoluto del juego colectivo que generaba el Barcelona.

2. Extremo Derecho: Es la posición en la que se dio a conocer mundialmente y la que más ha ocupado de manera intermitente en su carrera.Partiendo desde la banda derecha para trazar diagonales hacia el centro con su pierna izquierda ha sido su jugada más representiva, resultando en un volumen masivo de goles prácticamente los mismos que de falso 9 pero lograndolo en muchos más partidos.

3.  Mediapunta : En etapas de madurez futbolística, Messi ha retrasado su posición para actuar más como creador de juego. Aunque mantiene una capacidad goleadora sobresaliente para sus estándares y sublime para cualquier otro delantero, en estos roles su perfil se vuelca más hacia la asistencia y la distribución, disminuyendo ligeramente el volumen bruto de goles pero incrementando su influencia global en el juego.

```python
import pandas as pd
import plotly.express as px

# 1. Carga tus datos CSV (ajusta el nombre de tu archivo)
# df = pd.read_csv('goles_messi.csv')

# (Si tus columnas se llaman diferente, ajústalas abajo.
# Usaré datos de ejemplo basados en la imagen que subiste)
datos_ejemplo = {
    'Posición': ['Falso 9', 'Extremo Derecho', 'Mediapunta', 'Enganche', 'Extremo Izquierdo'],
    'Goles': [316, 316, 55, 16, 1]
}
df = pd.DataFrame(datos_ejemplo)

# 2. Crear la gráfica interactiva
# color='Goles' hará que cambie de amarillo a morado como en tu imagen
fig = px.bar(df,
             x='Posición',
             y='Goles',
             title='<b>Goles de Messi por Posición en el Campo</b>',
             text='Goles', # Muestra el número sobre la barra
             color='Goles',
             color_continuous_scale='Viridis_r') # Escala de colores similar a tu foto

# 3. Hacerla más atractiva (diseño limpio y tooltips al pasar el ratón)
fig.update_traces(
    textposition='outside',
    textfont_size=14,
    hovertemplate='<b>%{x}</b><br>Goles anotados: %{y}<extra></extra>' # Lo que sale al pasar el ratón
)

fig.update_layout(
    plot_bgcolor='rgba(0,0,0,0)', # Fondo transparente/blanco
    xaxis_title="Posición de Juego",
    yaxis_title="Cantidad de Goles",
    coloraxis_showscale=False, # Oculta la barra de colores lateral para que sea más limpia
    font=dict(size=14, color="#1e1e1e")
)

# Mostrarla en Colab para que interactúes con ella
fig.show()

# 4. ¡EL PASO CLAVE! Guardarla como un archivo web interactivo
fig.write_html("grafica_messi.html")
```

### Creación de un Dashboard Interactivo Premium (HTML)

El siguiente código lee los datos reales analizados de Messi y genera un archivo web interactivo (`dashboard_messi.html`) con un diseño altamente estético utilizando Tailwind CSS y Plotly, ideal para compartirlo con cualquier persona.

```python
import plotly.express as px
import pandas as pd
import numpy as np

# 1. Procesar los datos reales para asegurar consistencia
df_real = pd.read_csv("/content/data.csv", encoding="ISO-8859-1")
df_clean = df_real.dropna(subset=["Playing_Position"]).copy()
df_clean["Playing_Position"] = df_clean["Playing_Position"].astype(str).str.strip().str.upper()
df_clean = df_clean[df_clean["Playing_Position"] != "NAN"]

df_pos = df_clean["Playing_Position"].value_counts().reset_index()
df_pos.columns = ["Posición_Sigla", "Goles"]
df_pos["Porcentaje"] = (df_pos["Goles"] / df_pos["Goles"].sum() * 100).round(1)

posiciones_espanol = {
    'CF': 'Falso 9',
    'RW': 'Extremo Derecho',
    'SS': 'Mediapunta',
    'AM': 'Enganche',
    'LW': 'Extremo Izquierdo',
    'ST': 'Delantero Centro'
}
df_pos["Posición"] = df_pos["Posición_Sigla"].map(lambda x: posiciones_espanol.get(x, x))

# Calcular algunas métricas para el dashboard
total_goles = int(df_pos["Goles"].sum())
posicion_max = df_pos.iloc[0]["Posición"]
goles_max = int(df_pos.iloc[0]["Goles"])

# 2. Crear la gráfica interactiva optimizada
fig = px.bar(
    df_pos,
    x="Posición",
    y="Goles",
    color="Goles",
    color_continuous_scale="Viridis",
    text="Goles",
    hover_data={"Posición": True, "Goles": True, "Porcentaje": ":.1f%"}
)

fig.update_traces(
    textposition="outside",
    textfont=dict(size=12, color="#1e293b"),
    hovertemplate="<b>%{customdata[0]}</b><br><br>Goles: %{y}<br>Porcentaje: %{customdata[2]}%<extra></extra>",
    customdata=df_pos[["Posición", "Goles", "Porcentaje"]].values
)

fig.update_layout(
    margin=dict(l=20, r=20, t=20, b=20),
    plot_bgcolor="rgba(0,0,0,0)",
    paper_bgcolor="rgba(0,0,0,0)",
    xaxis_title=None,
    yaxis_title=None,
    coloraxis_showscale=False,
    font=dict(family="Inter, sans-serif", size=13, color="#334155")
)

fig.update_yaxes(showgrid=True, gridcolor="#e2e8f0", gridwidth=0.5)

# Obtener el div HTML de la gráfica
graph_div = fig.to_html(full_html=False, include_plotlyjs='cdn')

# 3. Construir la plantilla del Dashboard Web Premium
html_content = f"""<!DOCTYPE html>
<html lang="es">
<head>
    <meta charset="UTF-8">
    <meta name="viewport" content="width=device-width, initial-scale=1.0">
    <title>Goles de Messi por Posición</title>
    <!-- Tailwind CSS para diseño moderno -->
    <script src="https://cdn.tailwindcss.com"></script>
    <link href="https://fonts.googleapis.com/css2?family=Inter:wght@300;400;600;700&display=swap" rel="stylesheet">
    <style>
        body {{ font-family: 'Inter', sans-serif; }}
    </style>
</head>
<body class="bg-slate-50 min-h-screen py-8 px-4 flex items-center justify-center">
    <div class="max-w-4xl w-full bg-white rounded-3xl shadow-xl border border-slate-100 overflow-hidden">
        <!-- Encabezado -->
        <div class="bg-gradient-to-r from-blue-600 to-indigo-700 px-8 py-6 text-white">
            <span class="text-xs font-semibold tracking-wider uppercase bg-white/20 px-3 py-1 rounded-full">Análisis Interactivo</span>
            <h1 class="text-3xl font-bold mt-2">Goles de Lionel Messi por Posición</h1>
            <p class="text-blue-100 mt-1 text-sm">Visualiza el impacto goleador histórico según su rol táctico en la cancha.</p>
        </div>

        <!-- Tarjetas de Métricas Rápidas -->
        <div class="grid grid-cols-1 sm:grid-cols-3 gap-4 p-8 border-b border-slate-100 bg-slate-50/50">
            <div class="bg-white p-5 rounded-2xl border border-slate-100 shadow-sm">
                <p class="text-xs text-slate-400 font-semibold uppercase tracking-wider">Total Goles Analizados</p>
                <p class="text-2xl font-bold text-slate-800 mt-1">{total_goles}</p>
            </div>
            <div class="bg-white p-5 rounded-2xl border border-slate-100 shadow-sm">
                <p class="text-xs text-slate-400 font-semibold uppercase tracking-wider">Posición Más Letal</p>
                <p class="text-2xl font-bold text-indigo-600 mt-1">{posicion_max}</p>
            </div>
            <div class="bg-white p-5 rounded-2xl border border-slate-100 shadow-sm">
                <p class="text-xs text-slate-400 font-semibold uppercase tracking-wider">Récord en Posición</p>
                <p class="text-2xl font-bold text-slate-800 mt-1">{goles_max} Goles</p>
            </div>
        </div>

        <!-- Contenedor del Gráfico -->
        <div class="p-6 md:p-8">
            <div class="relative w-full overflow-hidden bg-white rounded-2xl">
                {graph_div}
            </div>
        </div>

        <!-- Pie de página informativo -->
        <div class="bg-slate-50 px-8 py-4 text-xs text-slate-400 flex justify-between items-center">
            <span>Datos actualizados</span>
            <span class="font-semibold text-slate-500">Interactúa pasando el cursor o tocando las barras</span>
        </div>
    </div>
</body>
</html>
"""

# 4. Guardar archivo final
with open("/content/dashboard_messi.html", "w", encoding="utf-8") as f:
    f.write(html_content)

print("¡Listo! Se ha creado el archivo 'dashboard_messi.html' en tu sección de archivos a la izquierda.")
```
